# DSPy Examples

```python
import dspy

qa = dspy.Example(question="What is 1+1?",answer="2")

print(qa)
print(qa.question)
print(qa.answer)
```

```python
qa = dspy.Example(article="This is an article",summary="This is a summary").with_inputs("article")

input_key = qa.inputs()
label_key = qa.labels()

print(qa)
print(input_key)
print(label_key)
```
