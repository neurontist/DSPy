```python
import os
import dspy

GEMINI_API_KEY = os.environ.get('GEMINI_API_KEY')

lm = dspy.LM('gemini/gemini-2.5-flash', api_key=GEMINI_API_KEY)
dspy.configure(lm=lm)

# Calling LM directly
lm("Say this is a test!", temperature=0.7)
lm(messages=[{"role":"user","content":"Say this is a test!"}])
```

### Using the LM with DSPy modules

```python
qa= dspy.ChainOfThought('question -> answer')

response = qa(question="How manu floors are in the castle David Gregory inherited?")
print(response.answer)
```

### Using multplie LMs

```python
dspy.configure(lm=dspy.LM('gemini/gemini-2.5-pro'))
response = qa(question="How manu floors are in the castle David Gregory inherited?")
print(response.answer)
```

```python
with dspy.context(lm=dspy.LM('gemini/gemini-2.5-flash', temperature=0.9, max_tokens=3000, cache=False)):
    response = qa(question="How manu floors are in the castle David Gregory inherited?")
    print(response.answer)
```

By default LMs in DSPy are cached. If you repeat the same call, you will get the same outputs. But you can turn off caching by setting cache=False.

```python
predict = dspy.Predict("question -> answer")
predict(question="What is 1 + 52?", config={"rollout_id":5, "temperature":0.5})
```

Every LM object maintains the history of its interactions, including inputs, outputs, token usage (and $$$ cost), and metadata.



```python
print("LM History Len",len(lm.history))
lm.history[-1].keys()
```
